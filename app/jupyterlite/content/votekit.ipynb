## Simulate an election

Two voter blocs and two slates of candidates, slate-Plackett-Luce ballots, then STV for two seats. See the [VoteKit docs](https://votekit.readthedocs.io) for everything else.

```python
from votekit.ballot_generator import BlocSlateConfig, slate_pl_profile_generator
from votekit.elections import FastSTV

config = BlocSlateConfig(
    n_voters=1000,
    slate_to_candidates={"A": ["A1", "A2", "A3"], "B": ["B1", "B2", "B3"]},
    bloc_proportions={"X": 0.6, "Y": 0.4},
    cohesion_mapping={"X": {"A": 0.8, "B": 0.2}, "Y": {"A": 0.3, "B": 0.7}},
)
config.set_dirichlet_alphas({"X": {"A": 1, "B": 1}, "Y": {"A": 1, "B": 1}})

profile = slate_pl_profile_generator(config)
FastSTV(profile, n_seats=2).get_elected()
```

```python
from votekit.plots import profile_fpv_plot

profile_fpv_plot(profile);
```
